```python
from pulp import *
```

### Definir o Problema

```python
problema = LpProblem("Wyndor Glass problem", LpMaximize)
```

### Variáveis de decisão

```python
x = Lpvariable('x', lowBound = 0)
y = Lpvariable('y', lowBound = 0)
```

### Função Objetivo

```python
problema += 3000*x + 5000*y
```

### Restrições

```python
problema += x + <=4
problema += 2*y + <=4
problema += 3*x + 2*y <=4
```

### Resolver

```python
problema.solve()

print('x =', value(x))
print('y =', value(y))
print('Solução Ótima', value(problema.objective))
```
